## C++ Programming

### Variables and Data Types

#### Boilerplate Code

```python
%%writefile boilerplate.cpp
#include<iostream> //preprocessor directive
using namespace std;

int main(){ //template or starting point
    cout << "xnocode" << endl;
}
```

```python
!g++ boilerplate.cpp -o boilerplate
!./boilerplate
```

#### Macros

```python
%%writefile macros.cpp
#include<iostream>
#define PI 3.1416
using namespace std;
int main(){ //template or starting point
    cout << PI << endl;
}
```

```python
!g++ macros.cpp -o macros
!./macros
```

#### Print Pattern

```python
%%writefile print-pattern.cpp
#include<iostream>
using namespace std;
int main(){
    cout <<"****"<< endl;
    cout <<"***"<< endl;
    cout <<"**"<< endl;
    cout <<"*"<< endl;
}
```

```python
!g++ print-pattern.cpp -o print-pattern
!./print-pattern
```

#### Variables

```python
%%writefile variables.cpp
#include<iostream>
using namespace std;
int main(){
    int a = 10;
    int b = 5;
    b = 50;
    int c;
    cout << a << endl;
    cout << b << endl;
    cout << c << endl;
}
```

```python
!g++ variables.cpp -o variables
!./variables
```

#### Size of Data Types

```python
%%writefile size-of-data-types.cpp
#include<iostream>
using namespace std;
int main(){
  int age = 25;
  char grade = 'A';
  bool isadult = true;
  cout << "size of int = " << sizeof(int) <<endl;
  cout << "size of char = " << sizeof(char) <<endl;
  cout << "size of bool = " << sizeof(bool) <<endl;
}
```

```python
!g++ size-of-data-types.cpp -o size-of-data-types
!./size-of-data-types
```

#### Input in C++

```python
%%writefile input.cpp
#include<iostream>
using namespace std;
int main(){
    int a, b;
    cin >> a >> b;
    cout << a + b;
}
```

```python
!g++ input.cpp -o input
!./input
```

```python
%%writefile age.cpp
#include<iostream>
using namespace std;
int main(){
    int age;
    cout << "Enter age: ";
    cin >> age;
    cout << "Your age is: " << age << endl;
}
```

```python
!g++ age.cpp -o age
!./age
```

```python
%%writefile average-mark.cpp
#include<iostream>
using namespace std;
int main(){
    int eng, math, sci;
    cin >> eng >> math >> sci;
    cout << "Average Mark: " << (eng + math + sci) / 3 << endl;
}
```

```python
!g++ average-mark.cpp -o average-mark
!./average-mark
```
